> randrange:
    
    - random.randrange(start,stop) 
    - stop number is exclusive 
     - if we put just the stop number in the (), it means start no. is 0


```python
r = random.randrange(5 , 11)
print(r)
```

> randint

    - random.randint(start,stop) 
    - here the stop number is INCLUSIVE
    - here we need to include the bottom of the range too

```python
r = random.randint(5 , 11)
print(r)
```

Adding number of guesses to the mix

```python
top_of_range = input("Type a random number ")
if top_of_range.isdigit():
    top_of_range = int(top_of_range)
    
    if top_of_range <= 0:
        print("Please type a number larger than 0 next time:(")
        quit()
else:
    print("Please type a number next time :(")
    quit()
    
# using the given number to generate a random number
random_number = random.randint(0 , top_of_range)
## if we print this,we will get a random number generated by the system
guesses = 0 # ADDING NUMBER OF GUESSES

# checking if the user can guess the random number
while True:
    guesses += 1 #everytime this loop is run, this score will be updated
    user_guess = input("Make a guess: ")
    if user_guess.isdigit():
        user_guess = int(user_guess)
    else:
        print("Please type a number next time :(")
        continue
        
    if user_guess == random_number:
        print("You got it!")
        break
    else:
        print("You got it WRONG :(")
        
#printing the no. of guesses
print("You got it in",guesses,"guesses")
```

---

Adding if the user is above or below the random number

```python
top_of_range = input("Type a random number ")
if top_of_range.isdigit():
    top_of_range = int(top_of_range)
    
    if top_of_range <= 0:
        print("Please type a number larger than 0 next time:(")
        return #exists the function
else:
    print("Please type a number next time :(")
    return #exists the function
    
# using the given number to generate a random number
random_number = random.randint(0 , top_of_range)
## if we print this,we will get a random number generated by the system
guesses = 0 

# checking if the user can guess the random number
while True:
    guesses += 1 
    user_guess = input("Make a guess: ")
    if user_guess.isdigit():
        user_guess = int(user_guess)
    else:
        print("Please type a number next time :(")
        continue
        
    if user_guess == random_number:
        print("You got it!")
        break
        
    else: #ADDING THE CONDITION IF > OR < than the random number
        if user_guess > random_number:
            print("You were above the number")
        else:
            print("You were below the number")
        
        
#printing the no. of guesses
print("You got it in",guesses,"guesses")
```

### Explaining BREAK and CONTINUE in while loop

| Keyword  | What it does?  | Effect  |
|----------|--------------|---------|
| `break`  | **Stops** the loop completely | Loop ends immediately |
| `continue` | **Skips** current iteration, but continues loop | Next iteration starts |

```python
# small example

while True:
    user_input = input("Guess a number (Type 'exit' to stop): ")

    if user_input == "exit":
        print("Game Over!")
        break  # Stops the game when user types 'exit'

    if not user_input.isdigit():
        print("Invalid input! Please enter a number.")
        continue  # Skips the rest and asks for input again

    print(f"You guessed: {user_input}")
```
