# 3. Rock Paper Scissors
---

- We will make a game of RPS and also keep a score of how many times we got it right v/s how many times the computer got it right

```python
import random
```

```python
user_wins = 0
computer_wins = 0

options = ["rock","paper","scissors"]
```

```python
while True:
    user_input = input("Type Rock/Paper/Scissors or Q to quit the game: ").lower()
    if user_input == "q":
        break # as this will immediately end the while loop
    
    if user_input not in options:
        continue #will reask the user to type r,p,s or q until they answer one of these
    
    random_number = random.randint(0, 2) 
    # rock : 0 , paper : 1 , scissors = 2
    
    computer_pick = options[random_number] #here we are indexing with random number
    print("Computer_picked", computer_pick + ".") #we used "+" not , because , will give spac
    
    #--- putting winning condition
    if user_input == "rock" and computer_pick == "scissors": #AND is used hence both conditions has to be true
        print("You Won!:)")
        user_wins += 1 # keeping the score
        
    elif user_input == "paper" and computer_pick == "rock":
        print("You Won! :)")
        user_wins += 1      
            
    elif user_input == "scissors" and computer_pick == "paper":
        print("You Won! :)")
        user_wins += 1
        
    else:
        print("YOU LOST! :(")
        computer_wins += 1

print("You won " ,user_wins, "times.")
print("The computer won ", computer_wins, "times.")
print("Goodbye!") 
```
