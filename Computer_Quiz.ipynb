# 1. Computer Quiz Game

```python
print("Welcome to my computer quiz!")
```

```python
# Asking the player if they want to play the game. If NO the game quits
playing = input("Do you want to play? ")

if playing.lower() != "yes":
    print("Exiting the game. Bye!")
else:
    print("Okay! Let's play :)")
    score = 0 

# Asking the first question of the quiz
answer = input("What does CPU stand for? ")

if answer.lower() == "central processing unit":
    print("Correct!")
    score = score + 1
else:
    print("Incorrect :(")
    

# Asking the second question of the quiz
answer = input("What does GPU stand for? ")

if answer.lower() == "graphics processing unit":
    print("Correct!")
    score += 1
else:
    print("Incorrect :(")


# Asking the third question of the quiz
answer = input("What does RAM stand for? ")

if answer.lower() == "random access memory":
    print("Correct!")
    score += 1
else:
    print("Incorrect :(")
    

# Asking the fourth question of the quiz
    ## using a different .lower() position in this one
answer = input("What does PSU stand for? ").lower()

if answer == " power supply":
    print("Correct!")
    score += 1
else:
    print("Incorrect :(")
    
print("You got " + str(score) + " questions correct")
```

```python
print("You got " + str((score/4)*100) + "%") 
```

### **Conclusion:**
- If you **only want to play when the user says "yes,"** keep `score = 0` inside the `else` block and prevent questions if they say "no."
  
- If you **always want to ask questions** regardless of the user's response, initialize `score = 0` outside the `if-else`.
